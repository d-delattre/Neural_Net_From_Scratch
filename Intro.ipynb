## 2. Mathematical Foundations

Understanding the mathematical principles behind neural networks is essential before implementing their architecture and learning algorithms. A neural network is fundamentally a parameterized mathematical function that transforms input data through a sequence of operations to produce predictions.

This section introduces the mathematical concepts required to understand and implement a feedforward neural network from scratch.

### 2.1. Mathematical Representation of a Neuron

An artificial neuron is a computational unit that receives one or more inputs, assigns a weight to each input, adds a bias, and applies an activation function.

For a neuron with \(n\) inputs, the weighted sum is defined as:

$$
z = \sum_{i=1}^{n} w_i x_i + b
$$

Where:

- \(x_i\) represents the \(i\)-th input feature.
- \(w_i\) represents the weight associated with that input.
- \(b\) represents the bias.
- \(z\) represents the pre-activation value.

The neuron subsequently applies an activation function:

$$
a = g(z)
$$

Where \(g\) is an activation function and \(a\) is the neuron's output.

#### Example

Consider a neuron with two inputs:

$$
x_1 = 2,\quad x_2 = 3
$$

Its corresponding weights and bias are:

$$
w_1 = 0.5,\quad w_2 = -0.2,\quad b = 0.1
$$

The weighted sum is:

$$
z = (2 \times 0.5) + (3 \times -0.2) + 0.1
$$

$$
z = 1.0 - 0.6 + 0.1 = 0.5
$$

Applying the ReLU activation function:

$$
g(z) = \max(0,z)
$$

Therefore:

$$
a = \max(0,0.5) = 0.5
$$

This illustrates the two fundamental operations performed by an artificial neuron: an affine transformation followed by an activation function.

### 2.2. Linear Algebra and Vectorization

Although individual neurons can be described using scalar operations, practical neural networks contain many neurons operating simultaneously.

Linear algebra provides an efficient mathematical framework for representing these computations.

Instead of processing each input independently, we represent inputs and weights as vectors and matrices.

For a neuron with \(n\) inputs, define:

$$
\mathbf{x} =
\begin{bmatrix}
x_1 & x_2 & \cdots & x_n
\end{bmatrix}
$$

$$
\mathbf{w} =
\begin{bmatrix}
w_1 & w_2 & \cdots & w_n
\end{bmatrix}
$$

The weighted sum can be expressed as:

$$
z = \mathbf{x}\mathbf{w}^{T}+b
$$

This is equivalent to the summation introduced previously, but expressed using matrix multiplication.

#### Matrix representation of a neural network layer

Consider a fully connected layer containing \(m\) neurons and receiving \(n\) input features.

The input vector is:

$$
\mathbf{x} \in \mathbb{R}^{1 \times n}
$$

The weight matrix is:

$$
W \in \mathbb{R}^{n \times m}
$$

The bias vector is:

$$
\mathbf{b} \in \mathbb{R}^{1 \times m}
$$

The layer computes:

$$
\mathbf{z} = \mathbf{x}W+\mathbf{b}
$$

And subsequently:

$$
\mathbf{a}=g(\mathbf{z})
$$

The output has dimensions:

$$
\mathbf{a} \in \mathbb{R}^{1 \times m}
$$

The dimensions are important because matrix multiplication requires the inner dimensions of its operands to match.

This matrix representation is the foundation of forward propagation.

### 2.3. Affine Transformations and Biases

The weighted sum of a neuron is technically an affine transformation rather than a purely linear transformation because it includes a bias.

An affine transformation has the general form:

$$
f(\mathbf{x})=W\mathbf{x}+\mathbf{b}
$$

The weight matrix determines how the input features are transformed, while the bias vector shifts the resulting values.

Without biases, the transformation is constrained to pass through the origin. Introducing biases provides additional flexibility and allows neural networks to represent more complex functions.

Both weights and biases are trainable parameters.

### 2.4. Activation Functions and Nonlinearity

An activation function transforms the pre-activation value of a neuron into its output.

Its primary purpose is to introduce nonlinearity into the neural network.

Without nonlinear activation functions, stacking multiple affine transformations would still produce a single affine transformation.

Consider two consecutive layers:

$$
f_1(\mathbf{x})=W_1\mathbf{x}+\mathbf{b}_1
$$

$$
f_2(f_1(\mathbf{x}))=W_2f_1(\mathbf{x})+\mathbf{b}_2
$$

Substituting the first transformation:

$$
f_2(f_1(\mathbf{x}))=
W_2W_1\mathbf{x}+W_2\mathbf{b}_1+\mathbf{b}_2
$$

This remains an affine transformation.

Consequently, adding more affine layers without nonlinear activation functions does not increase the fundamental expressive power of the resulting model.

Nonlinear activation functions allow neural networks to approximate complex relationships between input features and target variables.

#### Rectified Linear Unit (ReLU)

The Rectified Linear Unit is commonly used in hidden layers.

It is defined as:

$$
ReLU(z)=\max(0,z)
$$

Its derivative is:

$$
ReLU'(z)=
\begin{cases}
0 & \text{if } z < 0\\
1 & \text{if } z > 0
\end{cases}
$$

At \(z=0\), the derivative is undefined. Implementations conventionally assign a derivative of zero at this point.

ReLU is computationally inexpensive and helps mitigate the vanishing gradient problem associated with some saturating activation functions.

However, neurons can become inactive when their inputs consistently produce negative pre-activation values. This phenomenon is known as the dying ReLU problem.

#### Sigmoid

The sigmoid function is defined as:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

Its output lies between zero and one:

$$
0 < \sigma(z) < 1
$$

Its derivative can be expressed in terms of the function itself:

$$
\sigma'(z)=\sigma(z)(1-\sigma(z))
$$

The sigmoid function is useful for interpreting outputs as probabilities in binary classification, but its derivative becomes very small for large positive or negative inputs. This can contribute to vanishing gradients in deep networks.

#### Hyperbolic Tangent (Tanh)

The hyperbolic tangent activation function is:

$$
\tanh(z)=\frac{e^z-e^{-z}}{e^z+e^{-z}}
$$

Its output lies between negative one and positive one:

$$
-1 < \tanh(z) < 1
$$

Its derivative is:

$$
\tanh'(z)=1-\tanh^2(z)
$$

Unlike sigmoid, tanh produces zero-centered outputs, which can be beneficial during optimization.

#### Softmax

For multiclass classification, the Softmax function converts a vector of logits into a probability distribution.

For a vector containing \(K\) classes:

$$
Softmax(z_i)=
\frac{e^{z_i}}{\sum_{j=1}^{K}e^{z_j}}
$$

The resulting probabilities satisfy:

$$
0 < \hat{y}_i < 1
$$

And:

$$
\sum_{i=1}^{K}\hat{y}_i=1
$$

Softmax will be used in the output layer of our MNIST classifier.

For numerical stability, we will implement it using the equivalent expression:

$$
Softmax(z_i)=
\frac{e^{z_i-\max(\mathbf{z})}}
{\sum_{j=1}^{K}e^{z_j-\max(\mathbf{z})}}
$$

Subtracting the maximum logit does not change the resulting probabilities, but reduces the risk of numerical overflow.

### 2.5. Derivatives and Partial Derivatives

Training a neural network involves adjusting its parameters to minimize a loss function.

To determine how each parameter affects the loss, we need derivatives.

For a function:

$$
y=f(x)
$$

The derivative measures the instantaneous rate of change of the output with respect to the input:

$$
\frac{dy}{dx}
$$

Neural networks contain many parameters, so we frequently work with partial derivatives.

For a function depending on multiple variables:

$$
L=f(w_1,w_2,\ldots,w_n)
$$

The partial derivative with respect to parameter \(w_i\) is:

$$
\frac{\partial L}{\partial w_i}
$$

It measures how the loss changes when \(w_i\) changes while the other parameters are held constant.

These partial derivatives form the gradient of the loss function:

$$
\nabla L =
\begin{bmatrix}
\frac{\partial L}{\partial w_1}\\
\frac{\partial L}{\partial w_2}\\
\vdots\\
\frac{\partial L}{\partial w_n}
\end{bmatrix}
$$

The gradient points in the direction of steepest local increase of the function.

Therefore, moving in the opposite direction provides a way to reduce the loss.

### 2.6. The Chain Rule

The chain rule is one of the most important mathematical principles behind neural network training.

It allows us to calculate the derivative of a composite function.

Consider:

$$
y=f(g(x))
$$

Its derivative is:

$$
\frac{dy}{dx}
=
\frac{df}{dg}
\frac{dg}{dx}
$$

In other words, the derivative of a composition is obtained by multiplying the derivatives of its constituent functions.

Neural networks are compositions of functions. Each layer transforms the output of the preceding layer.

Consequently, the chain rule allows us to propagate derivatives backward through the network and determine how each parameter contributes to the final loss.

This process is called backpropagation.

The mathematical derivation of backpropagation will be developed in a dedicated section.

### 2.7. Loss Functions and Optimization

A neural network needs an objective function that quantifies the difference between its predictions and the actual target values.

This function is called the loss function.

For multiclass classification, we will use categorical cross-entropy:

$$
L=-\sum_{k=1}^{K}y_k\log(\hat{y}_k)
$$

Where:

- \(K\) is the number of classes.
- \(y_k\) represents the true target distribution.
- \(\hat{y}_k\) represents the predicted probability for class \(k\).

For a dataset containing \(N\) observations, the average loss is:

$$
J=-\frac{1}{N}
\sum_{i=1}^{N}
\sum_{k=1}^{K}
y_{ik}\log(\hat{y}_{ik})
$$

The objective of training is to find parameters that minimize this function.

One of the fundamental optimization algorithms used for this purpose is gradient descent.

Its parameter update rule is:

$$
\theta_{t+1}
=
\theta_t-\eta\nabla_\theta J(\theta_t)
$$

Where:

- \(\theta\) represents the model parameters.
- \(\eta\) represents the learning rate.
- \(\nabla_\theta J\) represents the gradient of the loss.
- \(t\) represents the current optimization iteration.

The learning rate determines the magnitude of each parameter update.

A learning rate that is too large can cause unstable optimization, while a learning rate that is too small can make convergence unnecessarily slow.

#### Summary

The mathematical foundations introduced in this section provide the theoretical basis for implementing a neural network.

The central concepts are:

1. **Affine transformations:** Combine inputs, weights and biases to produce pre-activation values.
2. **Linear algebra:** Represent computations efficiently through vectors and matrices.
3. **Activation functions:** Introduce the nonlinear properties required for learning complex relationships.
4. **Derivatives:** Measure how changes in parameters affect the loss.
5. **Chain rule:** Calculate gradients through compositions of mathematical functions.
6. **Loss functions:** Quantify prediction errors.
7. **Gradient descent:** Optimize model parameters by following the negative gradient.

These concepts will be used throughout the implementation of our multilayer perceptron.

## 3. Understanding Artificial Neural Networks

An artificial neural network (ANN) is a computational model composed of interconnected processing units called neurons. These neurons are organized into layers, with each layer transforming its input into a representation that is passed to the next layer.

Neural networks are designed to learn mathematical relationships between input variables and target outputs by adjusting their internal parameters during training.

In this project, we will implement a **Multilayer Perceptron (MLP)**, a type of feedforward neural network in which neurons from consecutive layers are fully connected.

### 3.1. Architecture of a Neural Network

A feedforward neural network typically consists of three types of layers:

- **Input layer:** Receives the input features and passes them to the first hidden layer.
- **Hidden layers:** Perform mathematical transformations and extract increasingly useful representations of the input data.
- **Output layer:** Produces the final predictions.

Information flows from the input layer toward the output layer without recurrent connections.

A neural network can contain multiple hidden layers, allowing it to learn increasingly complex representations.

#### General architecture

Consider a neural network with:

- One input layer containing \(n\) features.
- Two hidden layers containing \(h_1\) and \(h_2\) neurons.
- One output layer containing \(K\) neurons.

Its architecture can be represented as:

$$
n \rightarrow h_1 \rightarrow h_2 \rightarrow K
$$

Each layer performs an affine transformation followed by an activation function:

$$
Z^{[l]} = A^{[l-1]}W^{[l]} + b^{[l]}
$$

$$
A^{[l]} = g^{[l]}(Z^{[l]})
$$

Where:

- \(l\) denotes the current layer.
- \(A^{[l-1]}\) represents the activations from the preceding layer.
- \(W^{[l]}\) represents the weight matrix.
- \(b^{[l]}\) represents the bias vector.
- \(Z^{[l]}\) represents the pre-activation values.
- \(g^{[l]}\) represents the activation function.
- \(A^{[l]}\) represents the resulting activations.

The notation \(A^{[0]}\) represents the original input data.

The final layer produces:

$$
\hat{Y} = A^{[L]}
$$

Where \(L\) is the total number of trainable layers and \(\hat{Y}\) represents the model predictions.

### 3.2. The Multilayer Perceptron

A Multilayer Perceptron is a fully connected feedforward neural network.

Fully connected means that every neuron in a given layer receives information from every neuron in the preceding layer.

For example, consider a network with an input dimension of 4, a first hidden layer containing 3 neurons, a second hidden layer containing 2 neurons, and an output layer containing 2 neurons.

Its architecture is:

$$
4 \rightarrow 3 \rightarrow 2 \rightarrow 2
$$

The corresponding weight matrices have the following dimensions:

$$
W^{[1]} \in \mathbb{R}^{4 \times 3}
$$

$$
W^{[2]} \in \mathbb{R}^{3 \times 2}
$$

$$
W^{[3]} \in \mathbb{R}^{2 \times 2}
$$

The bias vectors have dimensions:

$$
b^{[1]} \in \mathbb{R}^{1 \times 3}
$$

$$
b^{[2]} \in \mathbb{R}^{1 \times 2}
$$

$$
b^{[3]} \in \mathbb{R}^{1 \times 2}
$$

The total number of trainable parameters is the sum of all weights and biases:

$$
P = \sum_{l=1}^{L}
\left(n_{l-1}n_l+n_l\right)
$$

Where:

- \(n_{l-1}\) is the number of neurons in the preceding layer.
- \(n_l\) is the number of neurons in the current layer.
- \(L\) is the number of trainable layers.

For the example above:

$$
P = (4 \times 3 + 3)
+ (3 \times 2 + 2)
+ (2 \times 2 + 2)
$$

$$
P = 15 + 8 + 6 = 29
$$

The network therefore contains 29 trainable parameters.

### 3.3. Understanding Information Flow

The forward propagation process describes how information moves through the network.

Each layer receives the activations produced by the previous layer, applies its weights and biases, and calculates new activations.

For a network containing two hidden layers and one output layer:

**First hidden layer:**

$$
Z^{[1]} = A^{[0]}W^{[1]}+b^{[1]}
$$

$$
A^{[1]} = g^{[1]}(Z^{[1]})
$$

**Second hidden layer:**

$$
Z^{[2]} = A^{[1]}W^{[2]}+b^{[2]}
$$

$$
A^{[2]} = g^{[2]}(Z^{[2]})
$$

**Output layer:**

$$
Z^{[3]} = A^{[2]}W^{[3]}+b^{[3]}
$$

$$
A^{[3]} = g^{[3]}(Z^{[3]})
$$

The final activation matrix \(A^{[3]}\) contains the network predictions.

For our MNIST classification problem, the output layer will use Softmax to produce probabilities corresponding to the ten possible digit classes.

### 3.4. Training: Learning the Parameters

Initially, the network does not know how to classify handwritten digits.

Its weights and biases are initialized with numerical values, and its initial predictions are generally inaccurate.

Training consists of repeatedly adjusting these parameters to reduce the difference between predicted and actual outputs.

The learning process involves four fundamental stages:

1. **Forward propagation:** Compute predictions using the current network parameters.
2. **Loss computation:** Evaluate the difference between predictions and target values.
3. **Backpropagation:** Calculate the gradients of the loss with respect to the network parameters.
4. **Parameter optimization:** Update weights and biases using an optimization algorithm.

This process is repeated over multiple iterations until the model learns useful patterns from the training data.

A complete pass through the entire training dataset is called an epoch.

### 3.5. Parameters and Hyperparameters

It is important to distinguish between model parameters and hyperparameters.

#### Model parameters

Parameters are values learned automatically during training.

For a fully connected neural network, these include:

- Weight matrices.
- Bias vectors.

The optimization algorithm updates these values to minimize the loss function.

#### Hyperparameters

Hyperparameters are configuration choices that influence the architecture or training process.

Examples include:

- Number of hidden layers.
- Number of neurons per layer.
- Learning rate.
- Batch size.
- Number of training epochs.
- Activation functions.
- Weight initialization strategy.

Unlike model parameters, hyperparameters are not directly learned through backpropagation.

They are selected and adjusted through experimentation and validation.


#### Summary

A Multilayer Perceptron consists of interconnected layers that transform input data through a sequence of affine and nonlinear operations.

The main concepts introduced in this section are:

1. **Network architecture:** The organization of neurons into input, hidden and output layers.
2. **Fully connected layers:** Every neuron receives information from all neurons in the preceding layer.
3. **Forward propagation:** The process of computing predictions through successive layers.
4. **Trainable parameters:** Weights and biases that are optimized during training.
5. **Hyperparameters:** Configuration choices that influence the architecture and learning process.
6. **Training process:** The iterative combination of forward propagation, loss computation, backpropagation and optimization.
